```python
import difflib, pathlib
import os

curfolder = os.getcwd()
parentfolder = os.path.join("..", curfolder)
```

```python
a = pathlib.Path(os.path.join(curfolder, "01_TS_extraction.py")).read_text().splitlines()
b = pathlib.Path(os.path.join(curfolder, "TS_extraction_prereg.py")).read_text().splitlines()

html = difflib.HtmlDiff(wrapcolumn=80).make_file(
    a, b, 
    fromdesc='NEW', 
    todesc='OLD',
    context=True,   # only show lines around changes (like nbdime)
    numlines=3      # how many context lines around each change
)

with open('01_TS_extraction_tracked.html', 'w', encoding='utf-8') as f:
    f.write(html)

print("Done! Open diff.html in your browser.")
```
