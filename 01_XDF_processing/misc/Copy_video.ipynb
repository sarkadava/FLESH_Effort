```python
import os
import glob

curfolder = os.getcwd()
videofolder = os.path.join(curfolder, '..', '..', 'video_only')
rawfolder = os.path.join(curfolder, '..', '00_raw_data')

videofiles = glob.glob(os.path.join(videofolder, '*.avi'))
print(videofiles)
```

We want to copy the videos into curfolder + respective folder of session

```python
bug = ['10_1', '11_1', '11_2', '13_1', '15_2', '16_1', '17_1']
```

```python
import shutil

reins = []

for videofile in videofiles:
    rein = False
    sessionid = os.path.basename(videofile).split('_')[0] + '_' + os.path.basename(videofile).split('_')[1]
    print(sessionid)
    # check if there is another file in videofiles that contains this sessionid
    for otherfile in videofiles:
        if otherfile != videofile:
            # get sessionid of otherfile
            other_sessionid = os.path.basename(otherfile).split('_')[0] + '_' + os.path.basename(otherfile).split('_')[1]
            if other_sessionid == sessionid:
                print(f'Found matching files: {videofile} and {otherfile}')
            #sessionfolder = os.path.join(curfolder, sessionid)
                rein = True
                reins.append(sessionid)
            # if not os.path.exists(sessionfolder):
            #     os.makedirs(sessionfolder)
            # shutil.copy(videofile, sessionfolder)
            # shutil.copy(otherfile, sessionfolder)
    if rein == False:
        print(f'No other match for {videofile}')

        # create folder if it does not exist
        sessionfolder = os.path.join(rawfolder, sessionid)
        if not os.path.exists(sessionfolder):
            os.makedirs(sessionfolder)


        if not os.path.exists(os.path.join(sessionfolder, sessionid + '-video.avi')):
            print(f'Copying {videofile} to {sessionfolder}')
            shutil.copy(videofile, os.path.join(sessionfolder, sessionid + '-video.avi'))

    #break

```

```python
# print unique instances in reins
reins = list(set(reins))
print(f'Unique reins: {reins}')
```
