```python
import os
import glob


curfolder = os.getcwd()
rawfolder = os.path.join(curfolder, '..', '00_rawdata')

avifiles = glob.glob(os.path.join(rawfolder, '*', '*.avi'), recursive=True)
```

```python
# find 50_1 in the list
samplefile = [f for f in avifiles if '50_1' in f][0]
```

```python
import subprocess

input_file = samplefile
output_file = "compressed_output.avi"

# Choose a high-quality AVI-compatible codec, like MJPEG or H.264 in AVI
# (libx264 works, but not all AVI players like it; MJPEG is more compatible)
command = [
    "ffmpeg",
    "-i", input_file,
    "-c:v", "libx264",        # or use "libx264" if you're okay with it in AVI
    "-qscale:v", "2",       # quality setting (1=best, 31=worst) for MJPEG
    "-c:a", "copy",         # copy audio as-is
    output_file
]

# Run the command
subprocess.run(command)
```
